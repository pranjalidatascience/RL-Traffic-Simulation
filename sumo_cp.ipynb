```python
import gymnasium as gym
import sumo_rl
from os import path
```

```python
# Use the package's built-in 4x4 grid example files for testing

base_path = "/Users/carol/anaconda3/lib/site-packages/sumo_rl"
net_file = path.join(base_path, "nets", "single-intersection", "single-intersection.net.xml")
route_file = path.join(base_path, "nets", "single-intersection", "single-intersection.rou.xml")

env = gym.make(
    'sumo-rl-v0', 
    net_file=net_file, 
    route_file=route_file, 
    out_csv_name='outputs/test_run', # Ensure the 'outputs' folder exists
    use_gui= False,
    num_seconds=100
)

obs, info = env.reset()
done = False
while not done:
    next_obs, reward, terminated, truncated, info = env.step(env.action_space.sample())
    done = terminated or truncated

env.close()
```
