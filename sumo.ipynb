```python
import os
import sys

# 1. Link SUMO 1.28.0
sumo_home = "/Library/Frameworks/EclipseSUMO.framework/Versions/Current/EclipseSUMO/share/sumo"
os.environ["SUMO_HOME"] = sumo_home
os.environ["PATH"] = f"{sumo_home}/bin:" + os.environ.get("PATH", "")
sys.path.insert(0, os.path.join(sumo_home, "tools"))

import gymnasium as gym
import sumo_rl
from stable_baselines3 import DQN

os.makedirs("outputs", exist_ok=True)
base_path = "/Users/pranjaligupta/anaconda3/lib/python3.11/site-packages/sumo_rl"
net_file = os.path.join(base_path, "nets", "single-intersection", "single-intersection.net.xml")
route_file = os.path.join(base_path, "nets", "single-intersection", "single-intersection.rou.xml")

# 2. Create the training environment
env = gym.make(
    "sumo-rl-v0",
    net_file=net_file,
    route_file=route_file,
    out_csv_name="outputs/ddqn_train",
    use_gui=False,
    num_seconds=10000,   # Longer episode gives the agent enough traffic to learn
    single_agent=True
)

# 3. Initialize DDQN (SB3's DQN uses Double Q-learning by default)
model = DQN(
    policy="MlpPolicy",
    env=env,
    learning_rate=1e-3,
    buffer_size=50000,
    learning_starts=500,
    batch_size=64,
    tau=1.0,
    gamma=0.99,
    train_freq=1,
    target_update_interval=500,
    exploration_fraction=0.2,
    exploration_final_eps=0.05,
    verbose=1
)

# 4. Train and save the model
model.learn(total_timesteps=20000)
model.save("ddqn_sumo_single")
env.close()
```

```python
# Load your trained DDQN model
model = DQN.load("ddqn_sumo_single")

obs, info = env.reset()
done = False
while not done:
    action, _ = model.predict(obs, deterministic=True)
    obs, reward, terminated, truncated, info = env.step(action)
    done = terminated or truncated
```
